```python
# Q7(e)
val1 = int(input("Enter First Number: "))
val2 = int(input("Enter Second Number: "))

def swap_message(a, b):
    print("Before Swap")
    print(a)
    print(b)
    temp = a
    a = b
    b = temp
    print("After Swap")
    print(a)
    print(b)

swap_message(val1, val2)
```

```python
# Q8(c)
counter = 0

def increment_counter():
    counter = counter + 1

try:
    increment_counter()
except UnboundLocalError as err:
    print("Error:", err)
```

```python
# Q8(d)
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

print(f"Before: {counter}")
increment_counter()
print(f"After: {counter}")
```
