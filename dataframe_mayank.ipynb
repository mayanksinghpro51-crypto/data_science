```python
# Q29: Add a partially indexed Series and then delete it

inventory_df = pd.DataFrame({
    "Product": ["Pen", "Book", "Bag", "Pencil", "Eraser"],
    "Price": [10, 50, 500, 5, 8]
})

stock_series = pd.Series(
    [20, 30],
    index=[1, 3]
)

inventory_df["Stock"] = stock_series

print("After adding Stock:")
print(inventory_df)

inventory_df = inventory_df.drop("Stock", axis=1)

print("\nAfter deleting Stock:")
print(inventory_df)
```
